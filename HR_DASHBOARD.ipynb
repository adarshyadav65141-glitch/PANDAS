```python
Q1 = df["MonthlyIncome"].quantile(0.25)
Q3 = df["MonthlyIncome"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower:", lower)
print("Upper:", upper)
```

```python
outliers = df[
    (df["MonthlyIncome"] < lower) |
    (df["MonthlyIncome"] > upper)
]

print("Number of outliers:", len(outliers))
```

```python
outliers["MonthlyIncome"].sort_values()
```
