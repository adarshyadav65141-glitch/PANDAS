```python
import pandas as pd
df=pd.read_excel("Day - 30 - Mobile Sales Data.xlsx")
```

```python
# Q1. Display the first 10 rows of the dataset.
print(df.head(10))
```

```python
# Q2. Find the number of rows and columns in the dataset.
print(df.shape)
```

```python
# Q3. Check the data types of all columns.
df.info()
```

```python
# Q4. Check whether the dataset contains any missing values
print(df.isnull().sum())
```

```python
# Q5. Find the number of unique brands and display their names.
print(df["Brand"].value_counts())
```

```python
# Q6. Count the number of transactions for each brand.
print(df.groupby("Brand")["Transaction ID"].count())
```

```python
# Q7. Find the total units sold by each brand.
print(df.groupby("Brand")["Units Sold"].sum())
```

```python
# Q8. Find the average price per unit for each brand.
print(df.groupby("Brand")["Price Per Unit"].mean().round(2))
```

```python
# Q9. Filter customers whose age is greater than 30.
print(df[df["Customer Age"] > 30])
```

```python
# Q10. Display all transactions where the payment method is UPI.
print(df[df["Payment Method"]=="UPI"])
```

```python
# Q11. Create a new column named "Sales Amount".
# Calculate Sales Amount = Units Sold × Price Per Unit.

df["Sales Amount"]=df["Units Sold"]*df["Price Per Unit"]
```

```python
# Q12. Calculate the total Sales Amount of all transactions.
print(df["Sales Amount"].sum())
```

```python
# Q13. Find which Brand has sold the highest number of Units.
brand_sales=df.groupby("Brand")["Units Sold"].sum()
print(brand_sales.idxmax())
print(brand_sales.max())
```

```python
# Q14. Find which City has sold the highest number of Units.
city_sales=df.groupby("City")["Units Sold"].sum()
print(city_sales.idxmax())
```

```python
# Q15. For each Payment Method, find:
# 1. Number of transactions     2. Total Units Sold     3. Total Sales Amount
payment_method=df.groupby("Payment Method").agg(Transcation=("Payment Method","count"),Total_Units_Solad=("Units Sold","sum"),
                                                Total_Sales_Amount=("Sales Amount","sum"))
print(payment_method)
```

```python
# Q16. Find the total Units Sold for each Mobile Model.
# Identify the highest-selling Mobile Model.
model_sels=df.groupby("Mobile Model")["Units Sold"].sum()
print(model_sels.idxmax())
```

```python

# Q17. Calculate the total Sales Amount for each Brand.
# Sort the result in descending order.
total_sale=df.groupby("Brand")["Sales Amount"].sum()
            .sort_values(ascending=False)
print(total_sale.map(lambda x: f"{x:,.2f}"))

```

```python
# Q18. Find the number of customers who gave each Customer Rating
# (1, 2, 3, 4, and 5).
rating=df.groupby("Customer Ratings").size()
print(rating)
```

```python

# Q19. Find the following statistics for Customer Age:
#       1. Minimum Age
#       2. Maximum Age
#       3. Average Age
#       4. Median Age

print("Minimum Age:", df["Customer Age"].min())
print("Maximum Age:", df["Customer Age"].max())
print("Average Age:", df["Customer Age"].mean().round(2))
print("Median Age:", df["Customer Age"].median())
```

```python
# Q20. Calculate the total Sales Amount for each Year and Brand.
Total_sales=df.groupby(["Year" , "Brand"])["Sales Amount"].sum()    
print(Total_sales.map(lambda x: f"{x:,.2f}"))
```
